# CROSS BORDER TAX DOCUMENT ASSISTANT

```python
import json, os
from google import genai
from google.genai import types

client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])

PROMPT = """Extract fields from this document as JSON only.
Keys: document_type, vendor_name, invoice_number, date, currency,
total_amount, account_number_last4, country, missing_fields (list).
Use null for anything not present."""

def extract(path: str, mime: str = "application/pdf") -> dict:
    with open(path, "rb") as f:
        data = f.read()
    resp = client.models.generate_content(
        model="gemini-3.5-flash-lite",  # check the current free-tier model name
        contents=[types.Part.from_bytes(data=data, mime_type=mime), PROMPT],
        config=types.GenerateContentConfig(response_mime_type="application/json"),
    )
    return json.loads(resp.text)

if __name__ == "__main__":
    print(extract("documents/invoice_brightwell-staffing-inc_BWS-2025-0104.pdf"))
```
